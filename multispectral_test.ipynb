```python
import cv2
import stackview
import tifffile as tf
```

```python
hella = tf.imread("rgbt.tiff")
```

```python
hella = tf.imread("image_rgbt_registered.tiff")
```

```python
stackview.switch(
{"red":   hella[:,:,0],
  "green":hella[:,:,1],
  "blue":hella[:,:,2],
  "lwir":hella[:,:,3],
},
colormap=["pure_red", "pure_green", "pure_blue", "inferno"],
toggleable=True, 
)
```
